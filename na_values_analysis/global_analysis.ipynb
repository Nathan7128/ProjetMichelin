```python
na_counts = df.isna().sum()
na_percent = ((na_counts / len(df)) * 100).round(2)
na_summary = pd.DataFrame({
    'na_count': na_counts,
    'na_percent': na_percent
}).sort_values('na_count', ascending=False)

na_summary
```

We note that for each tire, the 3 evaluated variables (temperature, pressure and cold pressure) have the same number of missing values.  
This probably means that these missing values are caused by issues associated with the tire sensors.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
for col in col_pressure:
    plt.hist(df[col].dropna(), bins=50, alpha=0.5, label=col)
plt.xlabel("Pressure value")
plt.ylabel("Count")
plt.title("Histogram of all pressure columns")
plt.legend()
plt.tight_layout()
plt.show()
```
